### Case: Score de Crédito dos Clientes

Você foi contratado por um banco para conseguir definir o score de crédito dos clientes. Você precisa analisar todos os clientes do banco e, com base nessa análise, criar um modelo que consiga ler as informações do cliente e dizer automaticamente o score de crédito dele: Ruim, Ok, Bom

Arquivos da aula: https://drive.google.com/drive/folders/1FbDqVq4XLvU85VBlVIMJ73p9oOu6u2-J?usp=drive_link

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
tabela = pd.read_csv('clientes.csv')

tabela.info()
```

```python
codigo_profissao = LabelEncoder()
tabela['profissao'] = codigo_profissao.fit_transform(tabela['profissao'])

codigo_credito = LabelEncoder()
tabela['mix_credito'] = codigo_credito.fit_transform(tabela['mix_credito'])

codigo_pagamento = LabelEncoder()
tabela['comportamento_pagamento'] = codigo_pagamento.fit_transform(tabela['comportamento_pagamento'])

```

```python
y = tabela['score_credito']
x= tabela.drop(columns=['id_cliente','score_credito'])

x_treino, x_teste, y_treino, y_teste = train_test_split(x, y, test_size=0.3)
```

```python
modelo_arvoredecisao = RandomForestClassifier()
modelo_knn = KNeighborsClassifier()

```

```python
modelo_arvoredecisao.fit(x_treino, y_treino)
modelo_knn.fit(x_treino, y_treino)
```

```python
previsao_arvoredecisao = modelo_arvoredecisao.predict(x_teste)
previsao_knn = modelo_knn.predict(x_teste)
```

```python
display(accuracy_score(y_teste, previsao_arvoredecisao))
display(accuracy_score(y_teste, previsao_knn))
```

```python
novos_clientes = pd.read_csv('novos_clientes.csv')



novos_clientes['profissao'] = codigo_profissao.fit_transform(novos_clientes['profissao'])


novos_clientes['mix_credito'] = codigo_credito.fit_transform(novos_clientes['mix_credito'])


novos_clientes['comportamento_pagamento'] = codigo_pagamento.fit_transform(novos_clientes['comportamento_pagamento'])

novos_clientes['mix_credito']

```

```python
nova_previsao = modelo_arvoredecisao.predict(novos_clientes)

display(nova_previsao)
```
